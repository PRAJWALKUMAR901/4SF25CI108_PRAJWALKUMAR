```python
man = man[['manga_id', 'title', 'type', 'score', 'scored_by','status', 'volumes', 'chapters', 'members','favorites', 'genres', 'demographics']]
print(man.head())
print("Shape:", man.shape)
```

```python
print( man.dtypes)
```

```python
plt.hist(man['score'].dropna(), bins=50)
plt.title("Score Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()

plt.boxplot(man['volumes'].dropna())
plt.title("Volumes Boxplot")
plt.show()

plt.boxplot(man['chapters'].dropna())
plt.title("Chapters Boxplot")
plt.show()
```

```python
man['score'] = man['score'].fillna(man['score'].mean())
man['volumes'] = man['volumes'].fillna(man['volumes'].median())
man['chapters'] = man['chapters'].fillna(man['chapters'].median())
```

```python
print("Missing values after filling:\n", man.isnull().sum())
```

```python
print("Top 10 popular mangas")
popular = man.sort_values(by='members', ascending=False)
print(popular[['title', 'members']].head(10))
```

```python
print("Average score")
type_score = man.groupby('type')['score'].mean()
print(type_score)
```

```python
print("Average members")
type_members = man.groupby('type')['members'].mean()
print(type_members)
```

```python
top10 = man.sort_values(by='members', ascending=False).head(10)
plt.bar(top10['title'], top10['members'])
plt.xticks(rotation=90)
plt.title("Top 10 Popular Manga")
plt.xlabel("Manga")
plt.ylabel("Members")
plt.show()
```

```python
plt.bar(type_score.index, type_score.values)
plt.title("Average Manga Score by Type")
plt.xlabel("Manga Type")
plt.ylabel("Average Score")
plt.show()
```
